```python

# Check missing values in every column

missing_data = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_data = missing_data[missing_data["Missing_Count"] > 0]

missing_data = missing_data.sort_values(
    by="Missing_Percentage",
    ascending=False
)

print("Missing Values Report:")
display(missing_data)
```

```python

# Check duplicate records

exact_duplicates = df.duplicated().sum()

key_duplicates = df.duplicated(
    subset=["order_id", "order_item_id"]
).sum()

print("Exact Duplicate Rows:", exact_duplicates)

print("Duplicate Order Items:", key_duplicates)
```

```python

# Check data types of all columns

print("Column Data Types:")

display(pd.DataFrame({
    "Column_Name": df.columns,
    "Data_Type": df.dtypes.astype(str).values
}))
```

```python
# Convert date columns into proper datetime format

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

print("Date columns converted successfully!")

display(df[date_columns].dtypes)
```

```python

# Check invalid numeric values

numeric_columns = [
    "price",
    "freight_value",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm",
    "average_review_score"
]

for col in numeric_columns:
    if col in df.columns:
        print(f"\n{col}")
        print("Minimum:", df[col].min())
        print("Maximum:", df[col].max())
        print("Negative values:", (df[col] < 0).sum())
```

```python

print("DATA DUPLICATE CHECK")

print("Exact duplicate rows:", df.duplicated().sum())

print("Duplicate order items:",
      df.duplicated(
          subset=["order_id", "order_item_id"]
      ).sum())

print("Total records:", len(df))
```

```python

numeric_columns = [
    "price",
    "freight_value",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm",
    "average_review_score"
]

for col in numeric_columns:
    print(f"\n--- {col} ---")
    print("Negative values:", (df[col] < 0).sum())
    print("Zero values:", (df[col] == 0).sum())
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
```

```python

# ==========================================
# SWYNEX - DATA CLEANING & PREPARATION
# Missing Values + Numeric Data Cleaning
# ==========================================

# 1. Store missing values before cleaning
missing_before = df.isnull().sum()

# 2. Fill missing product categories
category_columns = [
    "product_category_name",
    "product_category_name_english"
]

for col in category_columns:
    if col in df.columns:
        df[col] = df[col].fillna("unknown")

# 3. Fill only the small number of missing
# product dimension values using column median
dimension_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for col in dimension_columns:
    if col in df.columns:
        df[col + "_was_missing"] = df[col].isnull()
        df[col] = df[col].fillna(df[col].median())

# 4. Check negative numeric values
numeric_columns = [
    "price",
    "freight_value",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm",
    "average_review_score"
]

for col in numeric_columns:
    if col in df.columns:
        negative_count = (df[col] < 0).sum()
        print(f"{col}: {negative_count} negative values")

# 5. Keep legitimate missing information unchanged
# Delivery dates, reviews, payment details and
# unavailable product/customer/seller information
# will not be filled with invented values.

# 6. Generate missing values comparison
missing_after = df.isnull().sum()

missing_report = pd.DataFrame({
    "Before_Cleaning": missing_before,
    "After_Cleaning": missing_after,
    "Values_Handled": missing_before - missing_after
})

print("\nMISSING VALUES CLEANING REPORT")
display(missing_report[
    (missing_report["Before_Cleaning"] > 0) |
    (missing_report["After_Cleaning"] > 0)
])

print("\nDATA CLEANING STEP COMPLETED!")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python

# Check inconsistent categorical values

check_columns = [
    "order_status",
    "customer_geo_state",
    "seller_geo_state",
    "payment_types",
    "product_category_name_english"
]

for col in check_columns:
    print(f"\n--- {col} ---")
    print("Unique values:", df[col].nunique())
    print(df[col].value_counts(dropna=False).head(15))
```

```python

# Handle missing values safely

# Keep missing delivery dates as they may represent undelivered orders

# Fill missing product category with Unknown
df["product_category_name"] = df["product_category_name"].fillna("unknown")

df["product_category_name_english"] = (
    df["product_category_name_english"].fillna("unknown")
)

# Fill missing product review information is not appropriate,
# so preserve it as missing.

print("Missing values handled where appropriate!")

print("\nRemaining Missing Values:")
display(df.isnull().sum()[df.isnull().sum() > 0])
```

```python

# Final dataset validation

print("FINAL DATA VALIDATION")

print("\nTotal Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nExact Duplicate Rows:", df.duplicated().sum())

print("\nRemaining Missing Values:")
display(df.isnull().sum()[df.isnull().sum() > 0])

print("\nData Types:")
display(df.dtypes)
```
